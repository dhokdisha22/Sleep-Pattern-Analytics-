# Sleep Pattern Analysis — Exploratory Notebook

```python
import sys, os
sys.path.append(os.path.join('..', 'src'))
import pandas as pd
from data_cleaning import load_data, validate_data, clean_data
from analysis import compute_sleep_statistics, analyze_consistency, analyze_trend, identify_irregular_days
from visualization import plot_sleep_duration_trend, plot_duration_histogram, plot_quality_distribution
from ml_model import prepare_features, train_model, evaluate_model, save_model
from recommendations import generate_insights, generate_recommendations
```

## 1. Load & Validate

```python
raw_df = load_data('../data/raw/sleep_dataset.csv')
validate_data(raw_df)
```

## 2. Clean Data

```python
df = clean_data(raw_df)
df['date'] = pd.to_datetime(df['date'])
df.to_csv('../data/processed/cleaned_sleep_data.csv', index=False)
df.head()
```

## 3. Statistics & Consistency

```python
stats = compute_sleep_statistics(df)
consistency = analyze_consistency(df)
print(stats)
print(consistency)
```

## 4. Visualizations

```python
trend_df = analyze_trend(df)
plot_sleep_duration_trend(trend_df).show()
plot_duration_histogram(df).show()
plot_quality_distribution(df).show()
```

## 5. Machine Learning — Random Forest Classification

```python
X, y, encoder = prepare_features(df)
model, X_test, y_test = train_model(X, y)
metrics = evaluate_model(model, X_test, y_test, encoder)
print(metrics['classification_report'])
save_model(model, encoder)
```

## 6. Insights & Recommendations

```python
for i in generate_insights(df, stats, consistency):
    print('-', i)
print()
for r in generate_recommendations(stats, consistency):
    print('*', r)
```
